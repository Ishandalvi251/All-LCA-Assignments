## Exercise 1
Assign 5 elements in the dictionary named my_dict. This dictionary contains student's information. Add and delete student information from my_dict. Display student information.


```python
my_dict = {
    'name': 'ABC',
    'rollno': 34,
    'panel': 'B',
    'marks': 87,
    'city': 'Pune'
}
print("Original Dictionary:", my_dict)

# Add
my_dict['email'] = 'abc@gmail.com'
print("After Adding:", my_dict)

# Delete
del my_dict['city']
print("After Deleting:", my_dict)
```

## Exercise 3
Sort the elements of the dictionary using sorted() function.


```python
my_dict = {'name': 'ABC', 'panel': 'B', 'rollno': 34, 'marks': [65, 87, 67, 94]}

print("Sorted keys:", sorted(my_dict))
print("Sorted key-value pairs:", sorted(my_dict.items()))
```

## Exercise 4
Convert a dictionary to lists. Make one list of keys and other of values.


```python
my_dict = {'name': 'ABC', 'panel': 'B', 'rollno': 34}

keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print("Keys list:", keys_list)
print("Values list:", values_list)
```

## Exercise 5
Find the mean of values of dictionary.


```python
mydict = {'marks1': 23, 'marks2': 123, 'marks3': 43, 'marks4': 13, 'marks5': 39}

values = list(mydict.values())
mean_value = sum(values) / len(values)
print("Mean of values:", mean_value)
```

## Exercise 7
Calculate frequency of letters occurring in a given string.


```python
text = "programming"
freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print("Character Frequency:", freq)
```

## Exercise 9
Suppose a date is represented as a tuple (d,m,y). Write a program to create two date tuples with values taken from the user and find the number of days between the two dates as an output.


```python
from datetime import date

d1, m1, y1 = map(int, input("Enter date1 (d m y): ").split())
d2, m2, y2 = map(int, input("Enter date2 (d m y): ").split())

date1 = (d1, m1, y1)
date2 = (d2, m2, y2)

d_obj1 = date(date1[2], date1[1], date1[0])
d_obj2 = date(date2[2], date2[1], date2[0])

difference = abs((d_obj2 - d_obj1).days)
print("Number of days between the two dates:", difference)
```
